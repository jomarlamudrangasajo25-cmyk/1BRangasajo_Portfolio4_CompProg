(Interactive Smartwatch Activity Tracker)

In modern fitness technology, tracking physical activity is essential for monitoring health and lifestyle habits. Because walking paces and stride lengths vary from person to person, an interactive tracking system is needed to customize calculations. This program utilizes the input function paired  tocollect the user's specific walking data. The program then uses Iteration through a for loop to simulate a real-time tracking clock based on the inputted duration.

```python
walk_duration = int(input("Enter total walking time (in minutes): "))
steps_per_minute = int(input("Enter your walking pace (steps per minute): "))
stride_length = float(input("Enter your average stride length (in meters): "))
total_steps = 0
total_distance = 0.0
for minute in range(1, walk_duration + 1):
    total_steps = total_steps + steps_per_minute
    distance_this_minute = steps_per_minute * stride_length
    total_distance = total_distance + distance_this_minute
    print("Minute", minute, ": Steps = ",total_steps, ": Distance = ", total_distance, "meters")
print("Workout Summary")
print("Total Steps Taken:", total_steps, "steps")
print("Total Distance Traveled:", total_distance, "meters")
```
